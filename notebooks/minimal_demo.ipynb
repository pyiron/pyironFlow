# PyironFlow demo

`pyironflow` provides a visual scripting GUI to control the back-end `pyiron_workflow` model, which represents procedural workflows using a graph-and-node structure. Each node in the workflow has distinct inputs and outputs which can be connected by edges to form a workflow.

Click the "help" button in the GUI view to load a page on UI instructions (scroll to the "Basic usage" heading).

```python
import pyiron_workflow as pwf

from pyironflow import PyironFlow
```

```python
@pwf.as_function_node("signal")
def relu(x: float, bias:float = 0.0) -> float:
    return max(0.0, x - bias)

wf = pwf.Workflow("minimal_demo")
wf.n1 = relu(0.2)
wf.n2 = relu(-0.5)
wf.accumulate = pwf.std.Add(wf.n1, wf.n2)
wf.n3 = relu(wf.accumulate)
```

```python
pf = PyironFlow([wf])
pf.gui
```
